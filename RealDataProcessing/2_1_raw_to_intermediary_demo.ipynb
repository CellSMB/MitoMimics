# Imports

```python
from skimage import io
import numpy as np
import os
from pathlib import Path
```

```python
def rescale_array(array):
    # remove outliers
    p5 = np.percentile(array, 0.00001)
    p95 = np.percentile(array, 99.99999)
    
    # Perform scaling using p5 and p95
    scaled_array = (array - p5) / (p95 - p5)
    
    # Clip values to the range [0, 1]
    normalized = np.clip(scaled_array, 0, 1)    
    
    # Rescale to range [0, 256) and convert to uint8
    rescaled_array = (scaled_array * 256).astype(np.uint8)
    
    return rescaled_array
```

# Input variables

```python
input_folder = '2_1_input_raw_data/demo_raw_tiffstack/'
output_folder = '2_2_intermediary_data/demo_intermediary/'
```

```python
# create dict for new save locations for each file in the input folder

sample_folders = [os.path.join(input_folder, x) for x in os.listdir(input_folder)]
tif_locations_to_process = list()

for each in sample_folders:
    if '.ipynb_checkpoints' not in each:
        tif_locations_to_process.append(each)

in_out_dict = dict()

for input_path in tif_locations_to_process:
    filename = os.path.basename(input_path)                  # 'F5.tif'
    name_wo_ext = os.path.splitext(filename)[0]             # 'F5'
    output_path = os.path.join(output_folder, name_wo_ext)  # 'processed_outputs/F5'
    in_out_dict[input_path] = output_path

```

```python
# files that will be processed
in_out_dict
```

# Preprocessing

## NB: This will need to be modified depending on your input format. Re-write the reader as applicable

```python
for in_path, out_path in in_out_dict.items():

    # iterates over the input and output locations in the in_out_dict


    ## MODIFY THIS LINE SO THAT IT READS CORRRECTLY FOR YOUR MICROSCOPY DATA    
    raw_npy_array = io.imread(in_path)

    ## CHANGE THE INPUT TO a (t,x,y) 8 or 16bit numpy array
    raw_npy_array =  raw_npy_array.astype(np.uint16)

    ## RUN THE PREPROCESSING FUNCTION
    intermediary_processed_npy_array = rescale_array(raw_npy_array)

    ## SAVE
    np.save(out_path, intermediary_processed_npy_array)
```
