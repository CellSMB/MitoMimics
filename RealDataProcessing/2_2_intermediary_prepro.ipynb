# Imports

```python
from skimage import io
import numpy as np
import os
from pathlib import Path
import nibabel as nib
import skimage
```

# Input variables

```python
in_folder_path = '2_2_intermediary_data/demo_intermediary/'
out_folder_path = '2_3_processed_nifits/demo_preprocessed/'
```

# PREPROCESSING

## YOU WILL NEED TO MODIFY THE META DATA TO MATCH YOUR DATA

```python
image_id_counter = 1
```

```python
for current_id in os.listdir(in_folder_path):
    
    if current_id == '.ipynb_checkpoints':
        continue
    print(f'--------------------------------------------------')

    curr_id_no_tail = os.path.splitext(current_id)[0]             # 'F5'
    print(f'Proccesing image {curr_id_no_tail}')
    
    input_path = os.path.join(in_folder_path, current_id)

    print('Input_path:', input_path)

    current_raw_arr = np.load(input_path).astype(np.uint8)
    current_raw_arr = skimage.transform.resize(current_raw_arr, (current_raw_arr.shape[0], current_raw_arr.shape[1]*2,current_raw_arr.shape[2]*2), order=0, preserve_range=True, anti_aliasing=False)


    # prep with metadata for nnUNET 
    # WILL NEED TO MODIFY FOR YOUR CASE
    current_raw_arr = np.transpose(current_raw_arr, axes=[1,2,0]).astype(np.uint16)
    voxel_sizes = (0.054, 0.054, 0.054)
    affine = np.diag(voxel_sizes + (1.0,))  # (1.0) for homogeneous coordinates
    curr_img_nifti = nib.Nifti1Image(current_raw_arr, affine)
    curr_img_nifti.header.set_qform(affine, code=1)
    curr_img_nifti.header.set_sform(affine, code=1)
    curr_img_nifti.header.set_zooms(voxel_sizes)
    curr_img_nifti.header['xyzt_units'] = 10  # Units for micrometers in NIfTI
    curr_img_nifti.header['descrip'] = 'real microscopy data with 0.054 micron voxels'

    # out name
    img_fname = f'MIMIC_Ts_{str(image_id_counter).zfill(4)}_0000.nii.gz'
    print('Output file name')

    raw_file_name = os.path.join(out_folder_path, img_fname)
    nib.save(curr_img_nifti, raw_file_name)
```
